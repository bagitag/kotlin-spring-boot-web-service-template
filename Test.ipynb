## kandy

It is a plotting library.

```kotlin
%use kandy

posts.plot {
    x { it.userId }
    y { it.title.split("\\s+".toRegex()).size }
    geomBar(stat = "identity")
    theme {
        axis.text.x.angle = 45
        axis.text.x.hjust = 1
    }
}
```




````kotlin
%use ktor-client
%use kandy
%use dataframe
````

````kotlin
val users = http.get("https://jsonplaceholder.typicode.com/users").deserializeJson()
users
````

````kotlin
val usersDF = users.toDataFrame()
usersDF
````

````kotlin
val emailTLD = usersDF
    .add("emailTLD") { it.email.substringAfterLast('.') }
    .groupBy("emailTLD").count()
emailTLD
````

````kotlin
emailTLD.plot {
    layout.title = "User Email Top-level Domain Distribution"
    barsH {
        y(emailTLD) { axis.name = "TLD" }
        x(count) { axis.name = "count" }
        alpha = 0.75
        fillColor(emailTLD) {
            scale = categoricalColorHue()
        }
    }
}
````
